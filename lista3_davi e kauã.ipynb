```python
idades = []
idade = 0

print("Digite as idades (digite um número negativo para finalizar).")

while idade >= 0:
    try:
        idade = int(input("Digite uma idade: "))
        if idade >= 0:
            idades.append(idade)
    except ValueError:
        print("Entrada inválida. Por favor, digite um número inteiro.")

if idades:
    media = sum(idades) / len(idades)
    print(f"A média das idades digitadas é: {media:.2f}")
else:
    print("Nenhuma idade válida foi digitada.")
```

```python
maior_numero = float('-inf') # Initialize with a very small number

print("Digite 7 números para encontrar o maior.")
for i in range(7):
    try:
        numero = float(input(f"Digite o {i+1}º número: "))
        if numero > maior_numero:
            maior_numero = numero
    except ValueError:
        print("Entrada inválida. Por favor, digite um número.")

if maior_numero != float('-inf'):
    print(f"O maior número digitado é: {maior_numero}")
else:
    print("Nenhum número válido foi digitado.")
```

```python
tarefa_mais_longa = ""
maior_tamanho = 0

print("Digite os nomes de 4 tarefas:")
for i in range(4):
    nome_tarefa = input(f"Digite o nome da {i+1}ª tarefa: ")
    if len(nome_tarefa) > maior_tamanho:
        maior_tamanho = len(nome_tarefa)
        tarefa_mais_longa = nome_tarefa

if tarefa_mais_longa:
    print(f"A tarefa com o nome mais longo é: '{tarefa_mais_longa}' (com {maior_tamanho} caracteres).")
else:
    print("Nenhum nome de tarefa foi digitado.")
```

```python
positivos = 0
negativos = 0
zeros = 0

print("Digite 10 números:")
for i in range(10):
    try:
        numero = float(input(f"Digite o {i+1}º número: "))
        if numero > 0:
            positivos += 1
        elif numero < 0:
            negativos += 1
        else:
            zeros += 1
    except ValueError:
        print("Entrada inválida. Por favor, digite um número.")

print(f"Números positivos: {positivos}")
print(f"Números negativos: {negativos}")
print(f"Números iguais a zero: {zeros}")
```

```python

soma_valores = 0
print("Caixa Registradora Digital. Digite valores para somar (999 para finalizar):")

while True:
    try:
        valor = float(input("Digite um valor: "))
        if valor == 999:
            break
        soma_valores += valor
    except ValueError:
        print("Entrada inválida. Por favor, digite um número.")

print(f"A soma total dos valores é: {soma_valores:.2f}")
```

```python
senha_correta = "minhasenha123"
tentativas = 0
max_tentativas = 3

print(f"Sistema de Login. Você tem {max_tentativas} tentativas.")

while tentativas < max_tentativas:
    senha_digitada = input(f"Tentativa {tentativas + 1}/{max_tentativas}. Digite a senha: ")
    if senha_digitada == senha_correta:
        print("Login bem-sucedido! Bem-vindo(a)!")
        break
    else:
        print("Senha incorreta. Tente novamente.")
        tentativas += 1

else: # This else block executes if the while loop completes without a break
    print("Número máximo de tentativas excedido. Acesso bloqueado.")
```

```python
acumulador = 0
contador = 1

print("Iniciando acumulador. Somando números sequencialmente...")

while True:
    acumulador += contador
    print(f"Adicionando: {contador}, Soma atual: {acumulador}")
    if acumulador > 200:
        print("Soma excedeu 200. Encerrando o processo.")
        break
    contador += 1

print(f"Soma final: {acumulador}")
```
